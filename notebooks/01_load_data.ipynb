```python
df_clean = df.loc[df['Ridership_Total'] <= 300].copy()
```

```python
print(df_clean.shape)
```

```python
df_clean['Route_Date'] = pd.to_datetime(df_clean['Route_Date'], format="%m/%d/%Y %I:%M:%S %p")
df_clean['Route_Date'] = df_clean['Route_Date'].dt.normalize()
df_clean['Weekday'] = df_clean['Route_Date'].dt.day_name()
```

```python
df_clean.to_csv('../data/apc_clean.csv',index = False)
```

```python
print(df_clean['Route_Date'].min(), 'to' ,df_clean['Route_Date'].max())
print('Distint Dates:', df_clean['Route_Date'].nunique())

dates_per_weekday = df_clean.groupby('Weekday')['Route_Date'].nunique()
print(dates_per_weekday)

rows_per_date = df_clean.groupby('Route_Date').size()
print(rows_per_date.describe())
```

```python

all_days = pd.date_range("2026-01-01", "2026-10-27", freq="D")
missing = all_days.difference(df_clean['Route_Date'].unique())
print("Missing date(s):", missing)

rows_per_date = df_clean.groupby(['Route_Date', 'Weekday']).size().reset_index(name='rows')
print(rows_per_date.groupby('Weekday')['rows'].mean().round(0))

weekday_short = rows_per_date[
    (~rows_per_date['Weekday'].isin(['Saturday', 'Sunday'])) &
    (rows_per_date['rows'] < 3000)
]
print(weekday_short)
```

```python
import numpy as np
holiday_dates = weekday_short['Route_Date']
is_holiday = df_clean['Route_Date'].isin(holiday_dates)

df_clean['Day_Type'] = np.select(
    [is_holiday | (df_clean['Weekday'] == 'Sunday'),
     df_clean['Weekday'] == 'Saturday'],
    ['Sunday/Holiday', 'Saturday'],
    default='Weekday'
)

print(df_clean['Day_Type'].value_counts())


daily = df_clean.groupby(['Route_Date', 'Day_Type'])['Ridership_Total'].sum().reset_index()
print(daily.groupby('Day_Type')['Ridership_Total'].mean().round(0))
```
